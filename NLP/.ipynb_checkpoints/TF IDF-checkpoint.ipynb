```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus={
    "Thor eating pizza."
    "Loki is eating pasta."
    "Loki is eating pizza."
    "Iron Man ate pizza already."
    "Apple is announcing new iPhone tomorrow."
    "Tesla is announcing new Model 3 tomorrow."
    "Google is announcing new Pixel 6 tomorrow."
    "Microsoft is announcing new Surface tomorrow."
    "Amazon is announcing new Echo Dot tomorrow."
    "I am eating biryani."
    "You are eating grapes because you are a health freak."
}
```

```python
v=TfidfVectorizer()
v.fit(corpus)
transform_output=v.transform(corpus)
```

```python
print(v.vocabulary_)
```

```python
#Print the idf of each word
all_feature_name=v.get_feature_names_out()
for word in all_feature_name:
    indx=v.vocabulary_.get(word)
    idf_score=v.idf_[indx]
    print(f"{word}:{idf_score}")
```

## Applying Tf idf concept on dataset

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\bisha\Downloads\Ecommerce_data.csv")
```

```python
df.head()
```

```python
df['label'].value_counts()
```

```python
df['label']=df['label'].map({
    'Household':0,
    'Electronics':1,
    'Clothing & Accessories':2,
    'Books':3
})
```

```python
df.head(5)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(df.Text,df.label,random_state=42,test_size=0.2)
```

### Creation of model

```python
tf=TfidfVectorizer()
x_train_tf=tf.fit_transform(x_train)
x_test_tf=tf.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier

model=DecisionTreeClassifier()
model.fit(x_train_tf,y_train)
```

```python
from sklearn.metrics import classification_report
```

```python
y_pred=model.predict(x_test_tf)
print(classification_report(y_test,y_pred))
```
