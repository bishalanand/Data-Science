```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
url='https://raw.githubusercontent.com/jbrownlee/Datasets/refs/heads/master/airline-passengers.csv'
dataset=pd.read_csv(url,header=0,index_col=0,parse_dates=True)
```

```python
dataset.head()
```

```python
result=seasonal_decompose(dataset,model='multiplicative')
plt.rcParams.update({'figure.figsize':(10,10)})
result.plot()
plt.show()
```

### Performing Augmented Dickey Fuller

```python
from statsmodels.tsa.stattools import adfuller


result=adfuller(dataset['Passengers'])# pass only that column having time series column
adf_statistic = result[0]
p_value = result[1]
used_lag = result[2]
n_obs = result[3]
critical_values = result[4]  # dict with 1%, 5%, 10% critical values
icbest = result[5]  # Information criterion
```

```python
p_value
```

- p-value>0.05 failed to reject null hypothesise
- p-value<0.05 reject the null hypothesise
Here, p value is greater than 0.05 it means that we failed to reject that time series is not stationary
