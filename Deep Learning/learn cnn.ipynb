```python
import keras
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense,Dropout
from keras.layers import Conv2D,MaxPooling2D
from keras.layers import Flatten
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

we use mnsit data already present in keras containing hand written Digits.

```python
#Retaining the shape in 28*28
x_train=x_train.reshape(60000,28,28,1).astype('float32')
x_test=x_test.reshape(10000,28,28,1).astype('float32')
```

```python
#Making of cnn model
model=Sequential()

#First convolution layer
model.add(Conv2D(32,kernel_size=(3,3),activation='relu',input_shape=(28,28,1)))

#Second layer with pooling and dropout:
model.add(Conv2D(64,kernel_size=(3,3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2))) #Adding pooling layer of 2*2 for filtering
model.add(Dropout(0.25))
model.add(Flatten())

#Dense hidden layer , with dropout
model.add(Dense(128,activation='relu'))
model.add(Dropout(0.5))

#Outputlayer
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    x_train, y_train, 
    batch_size=128,     # how many samples per gradient update
    epochs=12,          # how many times the whole dataset is passed through
    validation_data=(x_test, y_test)  # evaluate on test set after each epoch
)
```
