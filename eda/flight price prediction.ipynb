## Flight price prediction(EDA + FEATURE ENGINEERING)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
test_df=pd.read_excel(r"C:\Users\bisha\Downloads\Test_set file attendant.xlsx")
test_df
```

```python
train_df=pd.read_excel(r"C:\Users\bisha\Downloads\Data_Train file attendant.xlsx")
train_df
```

```python
df=pd.concat([test_df,train_df])
```

```python
df.head()
```

```python
df.info()
```

```python
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Date']
```

```python
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['year']=df['year'].astype(int)
```

```python
df.info()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df.head()
```

```python
df['Arrival_hour']=df['Arrival_Time'].apply(lambda x:x.split(':')[0])
df['Arrival_minute']=df['Arrival_Time'].apply(lambda x:x.split(':')[1])
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_hour']=df['Arrival_hour'].astype(int)
df['Arrival_minute']=df['Arrival_minute'].astype(int)
```

```python
df['Dep_hour']=df['Dep_Time'].apply(lambda x:x.split(':')[0])
df['Dep_minute']=df['Dep_Time'].apply(lambda x:x.split(':')[1])
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df['Dep_hour']=df['Dep_hour'].astype(int)
df['Dep_minute']=df['Dep_minute'].astype(int)
```

```python
df.info()
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'1 stop':1, 'non-stop':0, '2 stops':2, '3 stops':3, '4 stops':4, 'nan':1})
```

```python
df.head()
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Duration']=df['Duration'].str.split('h')
```

```python
df['Duration']
```

```python
df['Duration_hour']=df['Duration'].apply(lambda x: x[0])
```

```python
df['Duration_minute']=df['Duration'].apply(lambda x:x[1] if len(x)>1 else 0)
```

```python
df['Duration_minute']=df['Duration_minute'].str.split('m')
```

```python
df['Duration_minute']
```

```python
df['Duration_minute'] = df['Duration_minute'].apply(
    lambda x: int(str(x[0]).replace('m', '').strip()) 
    if isinstance(x, list) and len(x) > 0 and str(x[0]).strip().replace('m', '').isdigit()
    else 0
)
```

```python
df.head()
```

```python
df['Duration_hour'] = df['Duration_hour'].apply(
    lambda x: int(str(x[0]).replace('m', '').strip()) 
    if isinstance(x, list) and len(x) > 0 and str(x[0]).strip().replace('m', '').isdigit()
    else 0
)
```

```python
df.info()
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
df.info()
```

```python
df['Destination'].unique()
```

```python
df['Destination']=le.fit_transform(df.Destination.values)
df['Destination']
```

```python
df['Airline']=le.fit_transform(df.Airline.values)
df['Additional_Info']=le.fit_transform(df.Additional_Info.values)
```
