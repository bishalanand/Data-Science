```python
text="""The paragraphos was introduced by scribes in the 4th century AD as a line placed beside a block of text.
These marginal symbols evolved into a pair of abbreviations. 
A stylized K, for kaput/caput or 'head', was placed at the beginning of new sections; a C, for capitulum or 'little head', which evolved into the pilcrow (¶), was placed at the beginning of subsections that modern readers would recognize as paragraphs.
In many English manuscripts in the Middle Ages, pilcrows were also placed inline between sentences.
Spaces were left at the beginning of paragraphs in early print documents for rubricators to hand-draw the pilcrows. 
These spaces were eventually left un-rubricated, leading to the modern tradition of indenting a paragraph. 
Using spacing alone to separate paragraphs eventually became standard.[4][5"""
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer()
tokenizer.fit_on_texts([text])
```

```python
len(tokenizer.word_index)
```

```python
len(text)
```

```python
leni = 0
for sentence in text.split('\n'):
    leni = max(leni, len(sentence))
    print(tokenizer.texts_to_sequences([sentence]))
```

```python
input_sequences=[]
for sentence in text.split('\n'):
    tok_sen=tokenizer.texts_to_sequences([sentence])[0]
    for i in range(1,len(tok_sen)):
        input_sequences.append(tok_sen[:i+1])
```

```python
leni
```

```python
max_len=max([len(x) for x in input_sequences])
```

```python
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded=pad_sequences(input_sequences,maxlen=max_len,padding='pre')
```

```python
padded
```

```python
x=padded[:,:-1]
```

```python
y=padded[:,-1]
```

```python
tokenizer.word_index
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=86)
```

```python
x.shape
```

```python
y.shape
```

```python
## Model building##
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,GRU,Dense
```

```python
model=Sequential()
model.add(Embedding(86,100,input_length=42))
model.add(GRU(150))
model.add(Dense(86,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
model.fit(x,y,epochs=100)
```

```python
model.summary()
```

```python
t_test=model.predict(x_test)
```
