```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\bisha\Downloads\All Codes (3)\Classification\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.TotalCharges=pd.to_numeric(df.TotalCharges,errors='coerce')
```

```python
df.info()
```

```python
df.dropna(how='any',inplace=True)
```

```python
df.info()
```

```python
y=df.Churn.values
```

```python
x=df.drop(['customerID','Churn'],axis=1)
```

```python
x.columns
```

```python
x=pd.get_dummies(x,columns=['gender','Partner','Dependents','PhoneService', 'MultipleLines', 'InternetService',
       'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
       'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract',
       'PaperlessBilling', 'PaymentMethod'],drop_first=True)
```

```python
x.head()
```

```python
x.columns
```

```python
x.info()
```

```python
x['gender_Male']=x['gender_Male'].astype(int)
```

```python
cols=['Partner_Yes', 'Dependents_Yes', 'PhoneService_Yes',
       'MultipleLines_No phone service', 'MultipleLines_Yes',
       'InternetService_Fiber optic', 'InternetService_No',
       'OnlineSecurity_No internet service', 'OnlineSecurity_Yes',
       'OnlineBackup_No internet service', 'OnlineBackup_Yes',
       'DeviceProtection_No internet service', 'DeviceProtection_Yes',
       'TechSupport_No internet service', 'TechSupport_Yes',
       'StreamingTV_No internet service', 'StreamingTV_Yes',
       'StreamingMovies_No internet service', 'StreamingMovies_Yes',
       'Contract_One year', 'Contract_Two year', 'PaperlessBilling_Yes',
       'PaymentMethod_Credit card (automatic)',
       'PaymentMethod_Electronic check', 'PaymentMethod_Mailed check']
for i in cols:
    x[i]=x[i].replace({True:'1',False:'0'}).astype(int)
```

```python
x.info()
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model=KNeighborsClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)*100
```

### Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
model_dt=DecisionTreeClassifier(max_depth=3)
```

```python
model_dt.fit(x_train,y_train)
```

```python
y_pred_dt=model_dt.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_dt)*100
```

### visualising the decision tree

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(10,12))
plot_tree(model_dt,filled=True,feature_names=x.columns,class_names=True)
plt.show()
```

### Pre-Prunning Decision Tree

```python
parameter={
    'criterion':['gini','entropy','log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['sqrt','log2',None]
}
```

```python
from sklearn.model_selection import GridSearchCV
treemodel=DecisionTreeClassifier()
```

```python
cv=GridSearchCV(treemodel,param_grid=parameter,cv=5,scoring='accuracy')
```

```python
cv.fit(x_train,y_train)
```

```python
cv.best_params_
```

```python
y_pred_cv=cv.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_cv)*100
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred_cv))
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model_rf=RandomForestClassifier()
model_rf.fit(x_train,y_train)
```

```python
y_pred_rf=model_rf.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_rf)*100
```
