## Removing stopwords

```python
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
sentence = "This is a sample sentence showing off the stopwords filtration."
words=word_tokenize(sentence)

new_sentence=[word for word in words if word.lower() not in stopwords.words('english')]
```

```python
print(new_sentence)
```

```python
from nltk import ngrams
def generate_ngrams(text,n):
    tokens=word_tokenize(text)
    n_grams=list(ngrams(tokens,n))
    return n_grams
```

```python
#example text
txt="N-grams are sequence of n items from a given sample of text or speech"
unigrams=generate_ngrams(txt,1)
digram=generate_ngrams(txt,2)
trigram=generate_ngrams(txt,3)
```

```python
print(unigrams)
print(digram)
```
