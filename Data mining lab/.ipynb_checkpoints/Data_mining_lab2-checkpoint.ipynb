### q1)

```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import pdist, squareform

data = pd.DataFrame({
    'Age': [25, 30, 35, 28, 40],
    'Income': [45, 60, 75, 50, 85],
    'Education': ['Bachelor', 'Master', 'PhD', 'Bachelor', 'Master']
})

numeric_data = data[['Age', 'Income']].values
distances = pdist(numeric_data, metric='euclidean')
dissimilarity_matrix = squareform(distances)

print("Dissimilarity Matrix:")
print(dissimilarity_matrix)
```

### q3)

```python
# Normalized size difference
size_diff = abs(1500 - 1800) / (3000 - 1000)

# Nominal (Type)
type_diff = 1  # Apartment ≠ House

# Ordinal condition
condition_map = {'Poor':1, 'Fair':2, 'Good':3, 'Excellent':4}
cond1 = condition_map['Good'] / 4
cond2 = condition_map['Excellent'] / 4
condition_diff = abs(cond1 - cond2)

# Binary symmetric (Garage)
garage_diff = 0  # both Yes

# Overall dissimilarity
overall = (size_diff + type_diff + condition_diff + garage_diff) / 4

print("Overall Dissimilarity:", overall)
```

### q4)

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from scipy.stats import entropy

docs = np.array([
    [3,0,2,1,4],
    [1,2,0,3,2],
    [2,1,1,2,3]
])

# Cosine similarity
cos_sim = cosine_similarity(docs)
print("Cosine Similarity Matrix:\n", cos_sim)

# Topics
A = np.array([0.4,0.3,0.2,0.1,0.0])
B = np.array([0.3,0.3,0.2,0.1,0.1])

print("KL(A||B):", entropy(A, B))
print("KL(B||A):", entropy(B, A))
```

### q5)

```python
import pandas as pd
import numpy as np
from word2number import w2n

# Create dataset
data = pd.DataFrame({
    'Name': ['John Doe', 'Jane Smith', 'Bob', 'Alice Johnson', np.nan],
    'Age': [25, 300, -5, 35, 28],
    'Email': ['john@email.com', 'jane.email@com', 'bob@company.org', 'alice@', 'charlie@test.com'],
    'Purchase_Date': ['2023-02-15', '2023/13/45', '2023-05-30', '2023-07-12', '2023-08-22'],
    'Amount': [150.50, 200.00, 'one hundred', 75.25, 300.00]
})

# 1. Handle missing names
data['Name'] = data['Name'].fillna('Unknown')

# 2. Fix invalid ages (18–100)
data['Age'] = data['Age'].astype(float)
data['Age'] = data['Age'].clip(lower=18, upper=100)

# 3. Standardize date format
data['Purchase_Date'] = pd.to_datetime(
    data['Purchase_Date'], errors='coerce'
)

new_amount = []

for i in data['Amount']:
    if isinstance(i, str):
        try:
            new_amount.append(w2n.word_to_num(i))
        except:
            new_amount.append(np.nan)
    else:
        new_amount.append(i)

data['Amount'] = pd.to_numeric(new_amount, errors='coerce')


print(data)
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 10000

df = pd.DataFrame({
    'customer_id': range(n_samples),
    'segment': np.random.choice(['A','B','C','D'], n_samples, p=[0.1,0.3,0.4,0.2])
})

# Simple random
srs = df.sample(1000)

# Stratified
stratified =df.groupby('segment', group_keys=False)\
               .apply(lambda x: x.sample(frac=0.1),include_groups=False)

# Systematic
systematic = df.iloc[::10]

# Reservoir sampling
reservoir = []
k = 1000
for i, row in df.iterrows():
    if i < k:
        reservoir.append(row)
    else:
        j = np.random.randint(0, i+1)
        if j < k:
            reservoir[j] = row
```

### q8)

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
import numpy as np

iris = load_iris()
X = iris.data

# Standardize
X_std = (X - X.mean(axis=0)) / X.std(axis=0)

# Covariance
cov = np.cov(X_std.T)

# Eigen
eigenvalues, eigenvectors = np.linalg.eig(cov)

# Sort
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]

# sklearn PCA
pca = PCA()
pca.fit(X_std)

print("Manual Eigenvalues:", eigenvalues)
print("Sklearn Eigenvalues:", pca.explained_variance_)
```

### q9)

```python
from sklearn.feature_selection import VarianceThreshold
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.random.rand(100,10)
y = np.random.rand(100)

# Variance threshold
vt = VarianceThreshold(0.01)
X_vt = vt.fit_transform(X)

# Correlation
corr = np.corrcoef(X.T, y)[:-1, -1]

# Forward selection
selected = []
remaining = list(range(10))

for _ in range(3):
    scores = []
    for f in remaining:
        model = LinearRegression()
        model.fit(X[:, selected + [f]], y)
        scores.append(model.score(X[:, selected + [f]], y))
    best = remaining[np.argmax(scores)]
    selected.append(best)
    remaining.remove(best)

print("Selected features:", selected)
```
