```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.metrics import r2_score
from sklearn.pipeline import Pipeline
```

```python
X=6*np.random.rand(200,1)-3#it makes data point lies between (-3,3)
y=0.8*X**2+0.9*X+2+np.random.randn(200,1) 
```

```python
plt.plot(X,y,'o')
plt.xlabel('x')
plt.ylabel('y')
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=3)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
r2_score_lr=r2_score(y_test,y_pred)
```

```python
r2_score_lr
```

```python
plt.plot(x_train,lr.predict(x_train),color='r')
plt.plot(X,y,'o')
plt.show()
```

## Applying polynomial linearregression

```python
poly=PolynomialFeatures(degree=2,include_bias=True)
x_train_trans=poly.fit_transform(x_train)
x_test_trans=poly.transform(x_test)
```

```python
lr2=LinearRegression()
lr2.fit(x_train_trans,y_train)
```

```python
y_pred2=lr2.predict(x_test_trans)
```

```python
r2_score(y_test,y_pred2)
```

```python
plt.plot(x_train_trans,lr.predict(x_train),color='r')
plt.plot(X,y,'o')
plt.show()
```
