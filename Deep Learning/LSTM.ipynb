```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset_train=pd.read_csv(r"C:\Users\bisha\Downloads\NSE-TATAGLOBAL.csv")
training_Set=dataset_train.iloc[:,1:2].values
```

```python
dataset_train.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
sc=MinMaxScaler(feature_range=(0,1))
training_Set_scaled=sc.fit_transform(training_Set)
```

```python
#creating a data structure with 60 timesteps and 1 output
x_train=[]
y_train=[]
for i in range(60,2035):
    x_train.append(training_Set_scaled[i-60:i,0])
    y_train.append(training_Set_scaled[i,0])
x_train,y_train=np.array(x_train),np.array(y_train)
```

```python
dataset_train.shape
```

```python
x_train.shape
```

```python
x_train=np.reshape(x_train,(x_train.shape[0],x_train.shape[1],1))#Increase dimension of neural network to 4
```

```python
x_train.shape
```

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import LSTM
from keras.layers import Dropout
```

```python
input_shape=(x_train.shape[1],1)
input_shape
```

```python
#Initialising the RNN
regressor=Sequential()
regressor.add(LSTM(units=50,return_sequences=True,input_shape=(x_train.shape[1],1)))
regressor.add(Dropout(0.2))
```

```python
#Adding second layer and some dropout regularisation
regressor.add(LSTM(units=50,return_sequences=True))
regressor.add(Dropout(0.2))
```

```python
#Adding third layer
regressor.add(LSTM(units=50,return_sequences=True))
regressor.add(Dropout(0.2))
```

```python
regressor.add(LSTM(units=50))

#Adding a output layer
regressor.add(Dense(units=1))

#Compiling the RNN
regressor.compile(optimizer='rmsprop',loss='mean_squared_error')
regressor.fit(x_train,y_train,epochs=10,batch_size=32)
```
