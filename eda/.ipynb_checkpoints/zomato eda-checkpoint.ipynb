```python
df=df.dropna()
```

```python
df.info()
```

```python
new_df=df
```

```python
country_names=new_df["Country Code"].value_counts().index
country_names
```

```python
country_val=new_df["Country Code"].value_counts().values
country_val
```

```python
plt.pie(country_val[:4],labels=country_names[:4],autopct="%1.2f%%")
```

observation:Zomato has maximum record or transaction are from country which have code =1,then from country having code=216,and then from 215

```python
new_df.columns
```

```python
ratings=new_df.groupby(["Aggregate rating","Rating color","Rating text"]).size().reset_index().rename(columns={0:'rating count'})
```

```python
ratings
```

```python
sns.barplot(data=ratings,x="Rating text",y="rating count",hue="Rating color",palette=['purple','red','orange','yellow','green','green'])
plt.title("Rating text vs count")
plt.show()
```

**observation:**
- Not rated count is very high
- maximum number of rating are good

```python
country_zerorating=new_df[new_df["Rating color"]=="White"].groupby("Country Code").size().reset_index().rename(columns={0:"count"})
country_zerorating
```
